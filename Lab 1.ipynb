```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show(img, title=''):
    # OpenCV is BGR; matplotlib expects RGB
    if img.ndim == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img)
    else:
        plt.imshow(img, cmap='gray', vmin=0, vmax=255)
    plt.title(title); plt.axis('off'); plt.show()
```

```python
img2=cv2.imread('/content/blue-mug_1101614-83164.jpg')
show(img2, 'Original')
hsv = cv2.cvtColor(img2, cv2.COLOR_BGR2HSV)

hsv_display = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

plt.imshow(hsv_display)
plt.title('HSV')
plt.axis('off')
plt.show()
gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
show(gray, 'Gray')
lower = np.array([100, 60, 30])
upper = np.array([130, 255, 255])
mask  = cv2.inRange(hsv, lower, upper)
show(mask, 'object mask')

result = cv2.bitwise_and(img2, img2, mask=mask)
show(result, 'Isolated Object')
#Hue thresholding is more robust to lighting changes because HSV separates color information from brightness. Hue mainly represents the actual color, while Value represents brightness, whereas RGB values change more directly when illumination changes.
```
